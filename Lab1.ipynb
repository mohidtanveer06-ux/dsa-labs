```python
array = [1,2,3,4,5,6,7]
print(array)
```

```python
if not array:
    print("Array is empty.")
else:
    largest_element = array[0]
    smallest_element = array[0]

    for element in array:
        if element > largest_element:
            largest_element = element
        elif element < smallest_element:
            smallest_element = element

    print(f"Largest element: {largest_element}")
    print(f"Smallest element: {smallest_element}")
```

```python

records = []
for i in range(1, 11):
    records.append(f"Student_{i}")
print("Initial Records:", records)
records.pop(2)
records.remove("Student_5")
print("\nTraversing records after removal:")
for student in records:
    print(f"Record: {student}")
records.append("New_Student_Final")
print("Final List:",records)
```

```python
n = int(input("How many patients are there? "))

patients = []

for i in range(n):
    print(f"\nEntering details for Patient {i+1}:")
    name = input("Enter Name: ")
    age = input("Enter Age: ")
    disease = input("Enter Disease: ")
    info = {"Name": name, "Age": age, "Disease": disease}
    patients.append(info)
print("\n--- Hospital Patient Records ---")
for p in patients:
    print(f"Name: {p['Name']}, Age: {p['Age']}, Disease: {p['Disease']}")
```

```python
numbers = []

print("Please enter 10 numbers:")
for i in range(10):
    val = int(input(f"Enter element {i+1}: "))
    numbers.append(val)
largest = numbers[0]
for x in numbers:
    if x > largest:
        largest = x
print(f"\nThe largest element is: {largest}")
```

```python
numbers = []

print("Please enter 10 numbers:")
for i in range(10):
    val = int(input(f"Enter element {i+1}: "))
    numbers.append(val)
search_num = int(input("\nEnter the number you want to count: "))
count = 0
for x in numbers:
    if x == search_num:
        count = count + 1
print(f"\nThe number {search_num} appears {count} times in the array.")
```
