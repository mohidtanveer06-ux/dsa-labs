```python
array = [1,2,3,4,5,6,7]
print(array)
```

```python
if not array:
    print("Array is empty.")
else:
    largest_element = array[0]
    smallest_element = array[0]

    for element in array:
        if element > largest_element:
            largest_element = element
        elif element < smallest_element:
            smallest_element = element

    print(f"Largest element: {largest_element}")
    print(f"Smallest element: {smallest_element}")
```

```python

records = []
for i in range(1, 11):
    records.append(f"Student_{i}")
print("Initial Records:", records)
records.pop(2)
records.remove("Student_5")
print("\nTraversing records after removal:")
for student in records:
    print(f"Record: {student}")
records.append("New_Student_Final")
print("Final List:",records)
```

```python

n = int(input("Enter the number of patients: "))
patients = []
for i in range(n):
    print(f"\nEntering details for Patient {i+1}:")
    name = input("Enter Name: ")
    age = input("Enter Age: ")
    disease = input("Enter Disease: ")
    patient_info = {"Name": name, "Age": age, "Disease": disease}
    patients.append(patient_info)
print("\n--- Hospital Patient Records ---")
for p in patients:
    print(f"Patient: {p['Name']} | Age: {p['Age']} | Disease: {p['Disease']}")
```
